# Generating waveforms for training data

Firstly, the generator tutorial from MLy will be used to generate different waveforms to then inject into noisy detector data.

```python
import numpy as np
import matplotlib.pyplot as plt

from mly.waveforms import *
from mly.projectwave import *
```

# CSG (low freq)

```python
csg_waveform_lowf_list = []

waveform_number = 100
f_min, f_max = 30, 99
d_min, d_max = 0.05, 0.5

for i in range(waveform_number):

    freq = f_min + np.random.rand()*(f_max - f_min) # Random values for frequency from 30Hz to 100Hz
    dur = d_min + np.random.rand()*(d_max - d_min) # Random values for duration from 0.2s to 1.0s

    #print(f"Frequency {freq}, Duration {dur}")
    csg_waveform = csg(
                        frequency = freq,
                        duration = dur,
                        fs=1024)

    csg_waveform_lowf_list.append(csg_waveform)
    
projected_csg_waveform_list = []

for csg_w in csg_waveform_lowf_list:

    w=projectWave(sourceWaveform = csg_w
                    ,detectors = 'HLV'
                    ,fs=1024
                    ,declination=None
                    ,rightAscension=None
                    ,polarisationAngle=None
                    ,time=0
                    ,outputFormat='datapod')

    projected_csg_waveform_list.append(w)

csg_injection_set = DataSet(projected_csg_waveform_list)
csg_injection_set.save('/home/jakub.rutkowski/project/waveform generation/waveform sets/csg_waveform_lowfreq.pkl')
```

# WNB

```python
wnb_waveform_list = []

waveform_number = 200
f_min, f_max = 30, 170
band_min, band_max = 10 , 100
d_min, d_max = 0.05, 0.5

for i in range(waveform_number):

    # Defining a random starting frequency.
    freq_start = f_min + np.random.rand()*(f_max - band_min - f_min)

    # Defining a random ending frequency within parameters.
    freq_end = freq_start + np.random.rand()*min(band_max , f_max - freq_start)

    # Random values for duration within parameters.
    dur = d_min + np.random.rand()*(d_max - d_min)

    #print(f"Frequencies {freq_start, freq_end}, Duration {dur}")
    wnb_waveform = WNB(duration = dur,
                            fs = 1024,
                            fmin = freq_start,
                            fmax = freq_end,
                            enveloped=True,
                            sidePad=None,
                            hrss = 1)

    wnb_waveform_list.append(wnb_waveform)
    
projected_wnb_waveform_list = []

for wnb_w in wnb_waveform_list:

    w=projectWave(sourceWaveform = wnb_w
                    ,detectors = 'HLV'
                    ,fs=1024
                    ,declination=None
                    ,rightAscension=None
                    ,polarisationAngle=None
                    ,time=0
                    ,outputFormat='datapod')

    projected_wnb_waveform_list.append(w)

wnb_injection_set = DataSet(projected_wnb_waveform_list)
wnb_injection_set.save('/home/jakub.rutkowski/project/waveform generation/waveform sets/wnb_waveform.pkl')
```
